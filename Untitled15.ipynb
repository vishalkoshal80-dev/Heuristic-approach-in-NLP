```python
def detect_sentiment(sentence):
  positive_words = [
      "good", "great", "excellent", "amazing",
      "happy", "love", "awesome", "fantastic"
  ]
  negative_words = [
      "bad", "worst", "terrible", "hate",
      "poor", "sad", "awful", "boring"
  ]

  pos_c = 0
  neg_c = 0

  words = sentence.lower().split()

  for word in words:
    if word in positive_words:
      pos_c += 1
    elif word in negative_words:
      neg_c +=1

  if pos_c > neg_c:
    return "positive"
  elif neg_c > pos_c:
    return "negative"
  else:
    return "Netural"

sentence = input("Enter a Sentence: ")

result = detect_sentiment(sentence)

print("Sentiment:" , result)
```

```python
def detect_intent(sentence):
  sentence = sentence.lower()

  if "hello" in sentence or "hi" in sentence:
    return "Greeting"

  elif "bye" in sentence or "goodbye" in sentence:
    return "Goodbye"

  elif "price" in sentence or "cost" in sentence:
    return "Product Price Query"
  elif "help" in sentence:
    return "Help Request"
  else:
    return "Unknown"

sentence = input("Enter a sentence: ")
result = detect_intent(sentence)
print("Intent: " , result)
```

```python
def detect_spam(message):
  spam_words = [
      "free", "winner", "prize", "offer",
      "urgent", "click", "money", "cash",
      "lottery", "congratulations"
  ]

  spam_count = 0

  words = message.lower().split()

  for word in words:
    if word in spam_words:
      spam_count += 1

  if spam_count >= 2:
    return "SPAM"
  else:
    return "NOT SPAM"

message = input("Enter a message")

result = detect_spam(message)

print("Result:", result)
```
